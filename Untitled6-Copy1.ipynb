```python
import rasterio
import numpy as np

# ================================
# input / output paths
# ================================
in_tif = r"E:\wenqu\2024_data\site2c\site2c_simulation_update1"
out_tif = r"D:\wenqu\2024_data\PFT_Map\site2c_M.tif"

# ================================
# selected bands and coefficients
# ================================
band_names = ['b74', 'b35', 'b21', 'b11', 'b62', 'b8', 'b10', 'b63', 'b6', 'b55',
              'b24', 'b50', 'b119', 'b49', 'b56', 'b7', 'b100', 'b117', 'b58',
              'b109', 'b51', 'b30', 'b108', 'b114', 'b52', 'b48', 'b112', 'b111',
              'b27']

coefficients = [-176.36549542, 207.51613489, 666.81607927, 167.47732022,
                -3.84433681, 108.92920972, -563.81061194, 185.55194082,
                -352.57352391, -279.52098561, -822.25412057, 627.64375433,
                -389.74086245, -418.91466271, -540.02834011, 526.22550223,
                52.28804455, 384.05867369, 454.73605904, 343.24575778,
                182.45173645, -523.63358838, 460.12425523, 454.6924985,
                598.36377844, -899.03464494, -582.79114337, -632.07745358,
                739.07755675]
intercept = 28.03047065
# convert band names like 'b74' -> 74
band_indices = [int(b[1:]) for b in band_names]

# ================================
# read tif and calculate weighted sum
# ================================
with rasterio.open(in_tif) as src:
    print("Raster band count:", src.count)
    print("Raster shape:", src.height, src.width)

    # initialize output raster
    result = np.zeros((src.height, src.width), dtype=np.float32)
    result += intercept

    # optional: build a combined valid mask
    valid_mask = np.ones((src.height, src.width), dtype=bool)

    for band_idx, coef in zip(band_indices, coefficients):
        print(f"Reading band {band_idx}, coefficient = {coef}")

        band = src.read(band_idx).astype(np.float32)

        # handle nodata if present
        if src.nodata is not None:
            valid_mask &= (band != src.nodata)

        result += band * coef

    # set invalid pixels to nodata
    out_nodata = -9999.0
    result[~valid_mask] = out_nodata

    # update metadata for single-band output
    out_meta = src.meta.copy()
    out_meta.update({
        "count": 1,
        "dtype": "float32",
        "nodata": out_nodata
    })

    with rasterio.open(out_tif, "w", **out_meta) as dst:
        dst.write(result, 1)

print("Done.")
print("Output saved to:", out_tif)
```
