```python
import os
import numpy as np
import rasterio
from rasterio.features import shapes
import geopandas as gpd
from shapely.geometry import shape

# ================================
# INPUT
# ================================
raster_path = r"E:\wenqu\2024_data\rgb\site1b_rgb.tif"
out_shp = r"E:\wenqu\2024_data\shp\site1b.shp"

# ================================
# READ RASTER AND BUILD MASK
# ================================
with rasterio.open(raster_path) as src:
    arr = src.read(1).astype(np.float32)
    nodata = src.nodata
    transform = src.transform
    crs = src.crs

    if nodata is not None:
        valid_mask = np.isfinite(arr) & (arr != nodata)
    else:
        valid_mask = np.isfinite(arr)

# 转成 0/1
valid_mask = valid_mask.astype(np.uint8)

# ================================
# VECTORIZE VALID AREA
# ================================
geoms = []
for geom, value in shapes(valid_mask, mask=valid_mask, transform=transform):
    if value == 1:
        geoms.append(shape(geom))

gdf = gpd.GeoDataFrame(geometry=geoms, crs=crs)

# ================================
# DISSOLVE TO ONE GEOMETRY
# ================================
merged = gdf.union_all()

# ================================
# REMOVE HOLES, KEEP ONLY OUTER BOUNDARY
# ================================
if merged.geom_type == "Polygon":
    outer_geom = type(merged)(merged.exterior)
elif merged.geom_type == "MultiPolygon":
    outer_geom = type(merged)([type(p)(p.exterior) for p in merged.geoms])
else:
    raise ValueError("Unexpected geometry type:", merged.geom_type)

out_gdf = gpd.GeoDataFrame(geometry=[outer_geom], crs=crs)

# ================================
# SAVE
# ================================
os.makedirs(os.path.dirname(out_shp), exist_ok=True)
out_gdf.to_file(out_shp)

print("Done.")
print("Output:", out_shp)
```

```python
import os
import numpy as np
import rasterio
from rasterio.features import shapes
import geopandas as gpd
from shapely.geometry import shape, Polygon, MultiPolygon

# ================================
# INPUT
# ================================
raster_path = r"E:\wenqu\2024_data\rgb\site1b_rgb.tif"
out_shp = r"E:\wenqu\2024_data\shp\site1b.shp"

# ================================
# READ RASTER
# ================================
with rasterio.open(raster_path) as src:
    arr = src.read(1)   # use band 1
    transform = src.transform
    crs = src.crs

    # background = 0, valid pixels are non-zero
    mask = (arr != 0) & np.isfinite(arr)

mask = mask.astype(np.uint8)

# ================================
# POLYGONIZE VALID AREA
# ================================
geoms = []
for geom, value in shapes(mask, mask=mask, transform=transform):
    if value == 1:
        geoms.append(shape(geom))

if len(geoms) == 0:
    raise ValueError("No valid non-zero pixels found.")

gdf = gpd.GeoDataFrame(geometry=geoms, crs=crs)

# ================================
# MERGE ALL POLYGONS
# ================================
merged = gdf.unary_union

# ================================
# REMOVE HOLES, KEEP ONLY OUTER PART
# ================================
if isinstance(merged, Polygon):
    outer_geom = Polygon(merged.exterior)

elif isinstance(merged, MultiPolygon):
    outer_geom = MultiPolygon([Polygon(p.exterior) for p in merged.geoms])

else:
    raise ValueError(f"Unexpected geometry type: {merged.geom_type}")

# ================================
# SAVE
# ================================
out_gdf = gpd.GeoDataFrame(geometry=[outer_geom], crs=crs)

os.makedirs(os.path.dirname(out_shp), exist_ok=True)
out_gdf.to_file(out_shp)

print("Done.")
print("Output:", out_shp)
```

```python
import os
import rasterio
import geopandas as gpd
from shapely.geometry import Polygon

# ================================
# INPUT
# ================================
raster_path = r"E:\wenqu\2024_data\rgb\site1b_rgb.tif"
out_shp = r"E:\wenqu\2024_data\shp\site1b.shp"
# ================================
# READ RASTER
# ================================
with rasterio.open(raster_path) as src:

    transform = src.transform
    width = src.width
    height = src.height
    crs = src.crs

    # four corners
    ul = transform * (0, 0)
    ur = transform * (width, 0)
    lr = transform * (width, height)
    ll = transform * (0, height)

# ================================
# CREATE POLYGON
# ================================
poly = Polygon([ul, ur, lr, ll])

gdf = gpd.GeoDataFrame(
    {"id": [1]},
    geometry=[poly],
    crs=crs
)

# ================================
# SAVE SHAPEFILE
# ================================
os.makedirs(os.path.dirname(out_shp), exist_ok=True)

gdf.to_file(out_shp)

print("Shapefile created:")
print(out_shp)
```
