```python
import rasterio
import pandas as pd
import numpy as np

asc_path = r"D:\wenqu\2024_data\SVC\site4_svc.asc"
csv_path = r"D:\wenqu\2024_data\SVC\site4_svc.csv"
data_started = False
rows = []

with open(asc_path, "r") as f:
    for line in f:
        line = line.strip()

        # SVC data table starts after this line
        if line.lower().startswith("data="):
            data_started = True
            continue

        if data_started:
            if line == "":
                continue
            parts = line.split()
            if len(parts) >= 2:
                rows.append(parts)

# Convert to DataFrame
df = pd.DataFrame(rows, columns=["wavelength_nm", "value"])
df = df.astype(float)

df.to_csv(csv_path, index=False)
print("Saved:", csv_path)
```

```python
import pandas as pd

asc_path = r"D:\wenqu\2024_data\SVC\site4_svc.csv"
csv_path = r"D:\wenqu\2024_data\SVC\site4_svc1.csv"

rows = []
data_started = False

with open(asc_path, "r") as f:
    for line in f:
        line = line.strip()

        if line.lower().startswith("data="):
            data_started = True
            continue

        if data_started:
            if not line:
                continue
            # split by ANY whitespace (spaces or tabs)
            parts = line.split()
            rows.append(parts)

# Convert to DataFrame
df = pd.DataFrame(rows)

# Convert all columns to numeric
df = df.apply(pd.to_numeric)

# Assign column names (adjust if needed)
df.columns = [
    "wavelength_nm",
    "target",
    "reference",
    "reflectance"
]

df.to_csv(csv_path, index=False)
print("Saved:", csv_path)
```
