```python
import os
import geopandas as gpd
import rasterio
from rasterio.mask import mask
from shapely.geometry import box

# Set your folder paths
shp_folder = r"D:\wenqu\shp\site6\single"
tif_folder = r"D:\wenqu\chapter1_2\chapter2\pft_spectra_origianl\PFT_Scale\projection"
output_folder = r"D:\wenqu\chapter1_2\chapter2\pft_spectra_origianl\PFT_Scale\projection\quadrant_clip\site6"

# Make sure output folder exists
os.makedirs(output_folder, exist_ok=True)

# Loop through each shapefile
for shp_file in os.listdir(shp_folder):
    if shp_file.endswith(".shp"):
        shp_path = os.path.join(shp_folder, shp_file)
        shapes = gpd.read_file(shp_path)
        shp_name = os.path.splitext(shp_file)[0]

        # Skip empty shapefiles
        if shapes.empty:
            print(f"[SKIP] Empty shapefile: {shp_file}")
            continue

        # Loop through each tif file
        for tif_file in os.listdir(tif_folder):
            if tif_file.endswith(".tif"):
                tif_path = os.path.join(tif_folder, tif_file)
                tif_name = os.path.splitext(tif_file)[0]

                with rasterio.open(tif_path) as src:
                    # Reproject shapes to raster CRS if needed
                    if shapes.crs != src.crs:
                        shapes_proj = shapes.to_crs(src.crs)
                    else:
                        shapes_proj = shapes

                    # Quick bbox overlap test (fast skip)
                    raster_bbox = box(*src.bounds)
                    shp_union = shapes_proj.unary_union
                    if not shp_union.intersects(raster_bbox):
                        print(f"[NO OVERLAP] {tif_file} × {shp_file} → skip")
                        continue

                    # Try mask/crop; if still no pixel overlap, catch and skip
                    try:
                        clipped_image, clipped_transform = mask(
                            src,
                            shapes_proj.geometry,
                            crop=True
                        )
                    except Exception as e:
                        print(f"[SKIP] {tif_file} × {shp_file} no overlap at pixel level ({e})")
                        continue

                    # If mask returns empty array (very rare), skip
                    if clipped_image.size == 0:
                        print(f"[SKIP] {tif_file} × {shp_file} produced empty clip")
                        continue

                    # Update metadata and save
                    clipped_meta = src.meta.copy()
                    clipped_meta.update({
                        "driver": "GTiff",
                        "height": clipped_image.shape[1],
                        "width": clipped_image.shape[2],
                        "transform": clipped_transform
                    })

                    out_tif = os.path.join(output_folder, f"{tif_name}_{shp_name}_clipped.tif")
                    with rasterio.open(out_tif, "w", **clipped_meta) as dest:
                        dest.write(clipped_image)

                    print(f"[SAVED] {out_tif}")
```

```python
import rasterio
from rasterio.warp import calculate_default_transform, reproject, Resampling
from rasterio.crs import CRS

# ---- paths ----
src_dem = r"D:\wenqu\2024_data\site7\site7_LiDAR\20240716_120429_1\processed\site7_dem_1.tif"
dst_dem = r"D:\wenqu\2024_data\site7\site7_LiDAR\20240716_120429_1\processed\site7_dem_1_utm6.tif"




# Target CRS: WGS 84 / UTM zone 5N (without using EPSG)
dst_crs = CRS.from_proj4("+proj=utm +zone=6 +datum=WGS84 +units=m +no_defs")

with rasterio.open(src_dem) as src:
    print("Source CRS:", src.crs)  # should be UTM zone 6N

    transform, width, height = calculate_default_transform(
        src.crs,
        dst_crs,
        src.width,
        src.height,
        *src.bounds
    )

    profile = src.profile.copy()
    profile.update({
        "crs": dst_crs,
        "transform": transform,
        "width": width,
        "height": height
    })

    with rasterio.open(dst_dem, "w", **profile) as dst:
        reproject(
            source=rasterio.band(src, 1),
            destination=rasterio.band(dst, 1),
            src_transform=src.transform,
            src_crs=src.crs,
            dst_transform=transform,
            dst_crs=dst_crs,
            resampling=Resampling.bilinear
        )

print("Done. Saved:", dst_dem)
```

```python
import pandas as pd
import geopandas as gpd
from shapely.geometry import Polygon
from pyproj import CRS, Transformer   

# ==== 你需要修改的两个路径 ====
csv_path = r"D:\wenqu\2024_data\csv\csv\site3_quadrant.csv"     # 你的 csv
shp_path = r"D:\wenqu\2024_data\shapefile\site3_quadrant.shp"  # 输出 shapefile

# ===== 读取 CSV =====
df = pd.read_csv(csv_path)

# 必须有 Name, Longitude, Latitude 三列
print(df.head())

# ===== 定义经纬度(WGS84) 和 UTM Zone 6N 的 CRS（用 proj4，不用 EPSG）=====
crs_wgs84 = CRS.from_proj4("+proj=longlat +datum=WGS84 +no_defs")
crs_utm6  = CRS.from_proj4("+proj=utm +zone=5 +datum=WGS84 +units=m +no_defs")

# 经纬度 → UTM Zone 6N 的转换器
transformer = Transformer.from_crs(crs_wgs84, crs_utm6, always_xy=True)

# ===== 按 Name 分组，一个 name 两行（两个对角点）=====
centers = []       # 保存中心点 (UTM 坐标)
names = []         # 保存 name

for name, group in df.groupby("Name"):

    if len(group) != 2:
        print(f"Warning: {name} 不等于 2 个点，跳过")
        continue

    lon1, lat1 = group.iloc[0]["Longitude"], group.iloc[0]["Latitude"]
    lon2, lat2 = group.iloc[1]["Longitude"], group.iloc[1]["Latitude"]

    # 经纬度 → UTM (米)
    x1, y1 = transformer.transform(lon1, lat1)
    x2, y2 = transformer.transform(lon2, lat2)

    # 计算中心点
    cx, cy = (x1 + x2) / 2, (y1 + y2) / 2

    centers.append((cx, cy))
    names.append(name)

# ===== 以中心点生成 50cm x 50cm 方形（单位：米）=====
half = 0.25  # 25 cm = 0.25 m

polygons = []
for cx, cy in centers:
    poly = Polygon([
        (cx - half, cy - half),
        (cx + half, cy - half),
        (cx + half, cy + half),
        (cx - half, cy + half),
        (cx - half, cy - half)
    ])
    polygons.append(poly)

# ===== 保存 shapefile，CRS 直接用 crs_utm6（proj4 / pyproj CRS），不用 EPSG 字符串 =====
gdf = gpd.GeoDataFrame({"Name": names}, geometry=polygons, crs=crs_utm6)
gdf.to_file(shp_path)

print("Shapefile saved:", shp_path)
```
